## implement a complete tokenizer class in Python.
### Step 1: Store the vocabulary as a class attribute for access in the encode and decode methods

### Step 2: Create an inverse vocabulary that maps token IDs back to the original text tokens

### Step 3: Process input text into token IDs

### Step 4: Convert token IDs back into text

### Step 5: Replace spaces before the specified punctuation
#### We can modify the tokenizer to use an <|unk|> token if it encounters a word that is not part of the vocabulary.

#### Furthermore, we add a token between unrelated texts.

```python
class tokenizer:

    def __init__(self,text):

        split_text = re.split(r'([,.:;?_!"()\']|--|\s)',text)
        final_text = [t for t in split_text if t.strip()]
        final_text = sorted(set(final_text))
        final_text.append('|unk|')
        final_text.append('|endoftext|')
        self.encode_vocab = [{token: index} for index,token in enumerate(final_text)]
        self.decode_vocab = [{index: token} for index, token in enumerate(final_text)]

    def encoder(self,t): # t is string
        self.encode = [self.encode_vocab[token] for token in t]
        return self.encode
    
    def decoder(self,t): # t is number
        self.decode = [self.decode_vocab[token_id] for token_id in t]
        return self.decode
```

```python
tokenizer_1 = tokenizer(text)

print(tokenizer_1.decoder([3, 11, 12, 13, 20, 21, 21, 0, 4]))
#print(tokenizer_1.encoder(['A','As', 'Be','At', '(', '!',')']))
```

```python
import importlib
import tiktoken

print("tiktoken version:", importlib.metadata.version("tiktoken"))
```

```python
tokenizer = tiktoken.get_encoding("gpt2")
```

```python
text = (
    "Hello, do you like tea? <|endoftext|> In the sunlit terraces"
     "of someunknownPlace."
)

integers = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
print(integers)
```

```python
strings = tokenizer.decode(integers)
print(strings)
```
